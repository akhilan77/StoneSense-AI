## Kidney Disease Classification from CT Images using Deep Learning

### ResNet18 + Grad-CAM Explainable AI

This notebook demonstrates the complete Deep Learning pipeline used in
StoneSense-AI for kidney CT image classification.

- Cyst
- Normal
- Stone
- Tumor
